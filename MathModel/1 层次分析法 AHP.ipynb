```python
import numpy as np
A = np.array([[1, 5, 3, 2], [1/5, 1, 1/2, 1/3], [1/3, 2, 1, 1/2], [1/2, 3, 2, 1]])
print(A.shape) #获取行，与列。或者用np.shape(array[])调用
n=A.shape[0]
```

```python
eig_val, eig_vec = np.linalg.eig(A) #获取特征值与特征向量
print(max(eig_val))
print(eig_vec)
```

```python
max_eig_val = max(eig_val)
CI = (max_eig_val-n)/(n-1)
print(CI)
RI = {1:0.00, 2:0.0001, 3:0.52, 4:0.89, 5:1.12,6:1.26, 7:1.36, 8:1.41, 9:1.46, 10:1.49,
      11:1.52, 12:1.54, 13:1.56, 14:1.58, 15:1.59}

CR = CI/RI[n]
print(CR)
# then compare CR with 0.10.
```

```python
#算数平均法
A_col_sum = np.sum(A, axis=0) #axis=0,按列求和
print(A_col_sum)
#归一化
stand_A = A / A_col_sum
print(stand_A)
#在内存中，A_col_sum 相当于被瞬间“复制拉伸”成了这样, 再逐个相除
#[ 2.03333333 11.  6.5  3.83333333]
#[ 2.03333333 11.  6.5  3.83333333]
#[ 2.03333333 11.  6.5  3.83333333]
#[ 2.03333333 11.  6.5  3.83333333]
weight = np.sum(stand_A, axis=1)/n
print(weight)
```

```python
#几何平均法
A_line_prod = np.prod(A, axis=1) #按行累×
print(A_line_prod)

A_line_prod_ = np.power(A_line_prod, 1/n)
weight = A_line_prod_ / np.sum(A_line_prod_)
print(weight)
```

```python
#特征值法
eig_val, eig_vec = np.linalg.eig(A) #获取特征值与特征向量
eig_max = max(eig_val)
max_idx = np.argmax(eig_val) #获取最大值索引
print(eig_vec.T[max_idx]) #列向量转置用行索引
weight = eig_vec.T[max_idx] / np.sum(eig_vec.T[max_idx])
print(weight)
print(type(weight[0]))
```
