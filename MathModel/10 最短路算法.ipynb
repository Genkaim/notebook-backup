```python
import numpy as np
import numpy as np

INF = 9999
# 节点 0~5，无向连通图，∞ 用 9999 占位
C = np.array([
    [   0,    7,  INF,    5,  INF,  INF],   # 0
    [   7,    0,    8,  INF,    2,  INF],   # 1
    [ INF,    8,    0,    3,  INF,    4],   # 2
    [   5,  INF,    3,     0,    6,  INF],   # 3
    [ INF,    2,  INF,    6,    0,    9],   # 4
    [ INF,  INF,    4,  INF,    9,    0],   # 5
], dtype=float)
```

```python
from scipy.sparse.csgraph import shortest_path

res = shortest_path(
    csgraph=C,
    indices=0,
    return_predecessors=True,
    directed=False,
)
print(res)
print(res[0][5])   # 节点1到节点5的最短距离
```

```python
path = res[1]
i = 5
print(f"{i} <-- ",end="")
while(i!=0 and path[i]!=-9999):
    i = path[i]
    if i!=0:
        print(f"{i} <-- ",end="")
    else:
        print(f"{i}")
```

```python
from scipy.optimize import milp, LinearConstraint, Bounds
#降维处理
#记录边
n = C.shape[0]
V = [(i, j, C[i][j])for i in range(n) for j in range(n) if 0<C[i][j]<9999]
#记录边权，系数矩阵
w = [w for (_, _, w) in V]
print(V)
print(w)

#系数部分
c = w
integrality = [1]*len(c)
bounds=Bounds(
    lb=[0]*len(c),
    ub=[1]*len(c)
)

#限制部分
# enumerate(E) 是 Python 内置函数，作用是遍历 E 时同时拿到"下标 k"和"元素本身"。
N = len(C[0])                     # 节点总数（你的图是 38）
m = len(V)                  # 边数
s = 0; t = 5   #起点终点
#出度：每一行代表第i个点（出发），对应在c的idx位置有边
c_out = np.zeros((N, m))
for idx, (i, j ,w) in enumerate(V):
    c_out[i][idx] = 1

#入度：每一行代表第j个点（进入），对应在c的idx位置有边
c_in = np.zeros((N, m))
for idx, (i, j ,w) in enumerate(V):
    c_in[j][idx] = 1

c_all = c_out-c_in
rhs = np.zeros(N)
rhs[s] = 1; rhs[t] = -1
con = LinearConstraint(c_all, rhs, rhs)

res = milp(
    c=c,
    bounds=bounds,
    constraints=con
)
print(res)
```
