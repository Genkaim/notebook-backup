```python
import numpy as np

# 节点顺序: S1..S7(0-6), B1..B16(7-22), A1..A15(23-37)
N = 38
INF = 9999
C = np.full((N, N), INF, dtype=float)
np.fill_diagonal(C, 0)

def idx(name):
    if name[0] == 'S': return int(name[1:]) - 1
    if name[0] == 'B': return 7 + int(name[1:]) - 1
    if name[0] == 'A': return 23 + int(name[1:]) - 1

edges = [
    # S -> B (铁路)
    ('S1','B8',1200), ('S2','B5',1150), ('S3','B9',690),
    ('S4','B10',170), ('S5','B11',160), ('S6','B12',320),
    # B -> A (公路)
    ('A1','B1',3), ('A3','B2',750), ('A4','B2',80), ('A4','B3',450),
    ('A4','B4',600), ('A5','B6',450), ('A6','B6',194), ('A8','B7',12),
    ('A8','B8',202), ('A7','B9',195), ('A8','B10',1100), ('A9','B11',480),
    ('A9','B12',42), ('A9','B13',70), ('A10','B15',300), ('A11','B16',10),
    # 管道 A1->A15 (双线)
    ('A1','A2',2), ('A2','A3',104), ('A3','A4',301), ('A4','A5',206),
    ('A5','A6',606), ('A6','A7',306), ('A7','A8',21), ('A8','A9',480),
    ('A9','A10',100), ('A10','A11',462), ('A11','A12',110), ('A12','A13',290),
    ('A13','A14',210), ('A14','A15',500),
    # S7 直连管道终点
    ('S7','A15',20),
]

for u, v, w in edges:
    C[idx(u), idx(v)] = w

```

```python
from scipy.sparse.csgraph import shortest_path

res = shortest_path(
    csgraph=C,
    indices=2,
    return_predecessors=True,
    directed=False,
)
print(res)
print(res[0][10])   # 节点1到节点10的最短距离
```

```python
path = res[1]
i = 10
print(f"{i} <-- ",end="")
while(i!=2 and path[i]!=-9999):
    i = path[i]
    if i!=2:
        print(f"{i} <-- ",end="")
    else:
        print(i)
```
