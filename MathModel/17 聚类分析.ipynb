```python
## 初始数据

X = [
    # 簇1
    [1.0, 1.2],
    [1.5, 1.8],
    [2.0, 1.5],
    [1.8, 2.3],
    [2.5, 2.0],
    [2.2, 2.8],
    [1.3, 2.5],

    # 簇2
    [7.0, 7.5],
    [7.5, 8.0],
    [8.0, 7.2],
    [8.5, 8.5],
    [9.0, 8.0],
    [8.2, 9.0],
    [7.8, 8.8],

    # 簇3
    [3.0, 8.0],
    [3.5, 8.5],
    [4.0, 7.8],
    [4.5, 8.8],
    [5.0, 8.2],
    [4.2, 9.3],
    [3.8, 9.0],

    # 离群点
    [6.0, 2.0],
    [10.0, 3.0]
]

```

```python
## 可视化

import matplotlib.pyplot as plt
import numpy as np


X = np.array(X)


plt.figure(figsize=(7,6))


plt.scatter(
    X[:,0],
    X[:,1],
    s=60
)


for i, (x,y) in enumerate(X):
    plt.text(
        x+0.1,
        y+0.1,
        str(i+1),
        fontsize=9
    )


plt.xlabel("x")
plt.ylabel("y")

plt.title("2D Sample Points")

plt.grid()

plt.show()
```

```python
from sklearn.cluster import KMeans
from sklearn.cluster import DBSCAN
from sklearn.cluster import AgglomerativeClustering

model = KMeans(
    n_clusters=4,
    init='k-means++'
)
kmeans_res = model.fit(X)

model = DBSCAN(
    eps=2,
    min_samples=1
)
dbscan_res = model.fit(X)

model = AgglomerativeClustering(
    n_clusters=4,
    linkage='single'
)
AgglomerativeClustering_res = model.fit(X)
```

```python
import matplotlib.pyplot as plt
import numpy as np


# 获取类别结果
kmeans_labels = kmeans_res.labels_

dbscan_labels = dbscan_res.labels_

agg_labels = AgglomerativeClustering_res.labels_


# 创建画布
plt.figure(figsize=(15,4))


# =====================
# K-means
# =====================
plt.subplot(1,3,1)

plt.scatter(
    X[:,0],
    X[:,1],
    c=kmeans_labels,
    s=60
)

# 绘制聚类中心
plt.scatter(
    kmeans_res.cluster_centers_[:,0],
    kmeans_res.cluster_centers_[:,1],
    marker='*',
    s=200
)

plt.title("K-means")
plt.xlabel("x")
plt.ylabel("y")
plt.grid()


# =====================
# DBSCAN
# =====================
plt.subplot(1,3,2)

plt.scatter(
    X[:,0],
    X[:,1],
    c=dbscan_labels,
    s=60
)

plt.title("DBSCAN")
plt.xlabel("x")
plt.ylabel("y")
plt.grid()


# =====================
# 层次聚类
# =====================
plt.subplot(1,3,3)

plt.scatter(
    X[:,0],
    X[:,1],
    c=agg_labels,
    s=60
)

plt.title("Agglomerative Clustering")
plt.xlabel("x")
plt.ylabel("y")
plt.grid()


plt.tight_layout()

plt.show()
```
