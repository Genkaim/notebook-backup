```python
import numpy as np
A = np.array([[90,4,35,15],[85,3,50,21],[92,4.5,40,19],[88,3.5,38,17]])
n, m = A.shape
```

```python
#极小型指标正向化
_A = A.copy().T #在新的array操作
# 正向化
mx = np.max(_A[1])
_A[1] = mx-_A[1]
print(_A)
```

```python
#中间型指标正向化
best = 40 
__A = _A.copy()
__A[2] = np.abs(__A[2]-40)
mx = np.max(__A[2])
_A[2] = 1 - np.abs(_A[2]-40)/mx
print(_A)
```

```python
#区间型正向化 16-18 best
a = 16; b = 18
mx = max(a-np.min(_A[3]), np.max(_A[3])-b)
print(mx)
for i in range(n):
    if _A[3][i] < a:
        _A[3][i] = 1-(a-_A[3][i]) / mx
    elif _A[3][i] <= b:
        _A[3][i] = 1
    else:
        _A[3][i] = 1-(_A[3][i]-b) / mx
print(_A)
```

```python
#归一化
A = _A.copy().T
print(A)
Aby = A.copy()*A.copy()
sm = np.sum(Aby, axis=0)

sm = np.sqrt(sm)
A = A/sm
print(A)
```

```python
Z_best = np.max(A, axis=0) #np.array([np.max(A.T[i]) for i in range(m)])
Z_bad = np.min(A, axis=0)
score = np.zeros(shape=(4,1)) #创建零矩阵
w = 1/m #权重
D_best = A.copy()
D_bad = A.copy()

#calculate
D_best = w * np.power((A-Z_best), 2)
D_best = np.power(np.sum(D_best, axis=1), 0.5)
print(D_best)

D_bad = w * np.power((A-Z_bad), 2)
D_bad = np.power(np.sum(D_bad, axis=1), 0.5)
print(D_bad)

score = D_bad/(D_best+D_bad)
print(score)
```
