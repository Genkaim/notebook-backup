```python
#这里使用Topsis法中，已经正向化，标准化的数据
import numpy as np
A = np.array([[0.50682714, 0.26726124, 0.36369648, 0.48507125],
[0.47867008, 0.80178373, 0.        , 0.        ],
 [0.51808997, 0.         ,0.72739297 ,0.48507125],
 [0.49556432, 0.53452248, 0.58191437 ,0.72760688]])
print(A)
n = A.shape[0]
```

```python
#归一化
sm = np.sum(A, axis=0)
A = A/sm
print(A)
```

```python
A = (-1) * A * np.log(A, out=np.zeros_like(A), where= A>0) * 1/(np.log(n))
# np.zeros_like(A) 的参数就是一个参考数组，
# NumPy 会根据它创建一个形状（shape）和数据类型（dtype）相同、元素全为 0 的新数组。
print(A)
Dj = np.sum(A,axis=0)
Wj = 1-Dj
Wj = Wj/np.sum(Wj)
print(Wj)
```
