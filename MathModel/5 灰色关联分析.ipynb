```python
import numpy as np
A = np.array([[80,70,85,75,],[70,65,78,80],[90,75,80,85],[75,60,70,80]])
print(A)
```

```python
#已经正向化，直接统一量纲
mean = np.mean(A, axis=0)
print(mean)
A = A / mean
print(A)
```

```python
#母序列
mother_array = np.max(A, axis=0)
print(mother_array)
A = np.abs(mother_array - A)
print(A)
```

```python
# 参数们
rho = 0.5
a = np.min(np.min(A,axis=0))
b = np.max(np.max(A,axis=0))
print(a, b)
```

```python
A = (a + rho*b) / (A + rho*b)
print(A)
ans = np.mean(A, axis=1)
print(ans)
```
